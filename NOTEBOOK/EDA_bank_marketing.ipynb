```python
# Diccionario para convertir los meses españoles a números
meses = {
    "enero": "01",
    "febrero": "02",
    "marzo": "03",
    "abril": "04",
    "mayo": "05",
    "junio": "06",
    "julio": "07",
    "agosto": "08",
    "septiembre": "09",
    "octubre": "10",
    "noviembre": "11",
    "diciembre": "12"
}

# Función para convertir las fechas españolas
def convertir_fecha(fecha):
    if pd.isna(fecha):
        return pd.NaT

    partes = fecha.strip().lower().split("-")

    if len(partes) != 3 or partes[1] not in meses:
        return pd.NaT

    dia, mes, anio = partes

    return pd.to_datetime(
        f"{dia}-{meses[mes]}-{anio}",
        format="%d-%m-%Y",
        errors="coerce"
    )

# Creamos una columna con las fechas convertidas
bank_limpio["date_convertida"] = bank_limpio["date"].apply(convertir_fecha)

# Comprobamos los resultados
print("Tipo de dato:")
print(bank_limpio["date_convertida"].dtype)

print("\nPrimeras fechas convertidas:")
print(bank_limpio[["date", "date_convertida"]].head(10))

print("\nValores nulos después de la conversión:")
print(bank_limpio["date_convertida"].isna().sum())
```

Tras la transformación, se mantuvieron los 248 valores faltantes originales, representados como `NaT`. Esta conversión permitirá realizar análisis temporales por meses y años.

```python
# ANÁLISIS: CORRELACIONES CON LA CONTRATACIÓN

# Creamos una variable binaria para representar la contratación
bank_final["contrato"] = (bank_final["y"] == "yes").astype(int)

# Seleccionamos variables numéricas relevantes
variables_numericas = [
    "age",
    "Income",
    "Kidhome",
    "Teenhome",
    "NumWebVisitsMonth",
    "duration",
    "campaign",
    "previous",
    "emp.var.rate",
    "cons.price.idx",
    "cons.conf.idx",
    "euribor3m",
    "nr.employed",
    "contrato"
]

# Calculamos las correlaciones
correlaciones = bank_final[variables_numericas].corr()

# Ordenamos las correlaciones con la contratación
correlacion_contratacion = correlaciones["contrato"].drop(
    "contrato"
).sort_values()

print("Correlaciones con la contratación:")
print(correlacion_contratacion.round(3))

# Representación gráfica
plt.figure(figsize=(9, 7))

colores = [
    "#2E8B79" if valor >= 0 else "#D97732"
    for valor in correlacion_contratacion
]

plt.barh(
    correlacion_contratacion.index,
    correlacion_contratacion.values,
    color=colores
)

plt.axvline(x=0, color="black", linewidth=0.8)
plt.title("Correlación de variables numéricas con la contratación")
plt.xlabel("Coeficiente de correlación de Pearson")
plt.tight_layout()
plt.show()
```

### Análisis de correlaciones entre variables numéricas y contratación

Se calcularon los coeficientes de correlación de Pearson entre distintas variables numéricas y la contratación del depósito bancario, representada mediante una variable binaria (`contrato`).

La duración de la llamada (`duration`) presenta la mayor correlación positiva (0,405), seguida del número de contactos anteriores (`previous`), con un valor de 0,233.

Por otro lado, las variables económicas `nr.employed` (-0,356), `euribor3m` (-0,312) y `emp.var.rate` (-0,298) muestran las correlaciones negativas más destacadas.

En contraste, variables como los ingresos (`Income`), el número de hijos (`Kidhome`, `Teenhome`) y las visitas mensuales a la web (`NumWebVisitsMonth`) presentan correlaciones lineales prácticamente nulas.

La edad también muestra una correlación lineal reducida (0,029), aunque el análisis por grupos de edad reveló diferencias relevantes. Esto pone de manifiesto que las correlaciones lineales no permiten identificar todas las relaciones posibles entre variables.

Es importante destacar que la duración de la llamada no está disponible antes del contacto comercial, por lo que no debería utilizarse como variable predictora para seleccionar clientes antes de llamarlos.

Asimismo, las correlaciones observadas no implican causalidad y deben interpretarse dentro del contexto del análisis exploratorio.

Estas diferencias sugieren que las características sociodemográficas pueden resultar útiles para identificar segmentos de interés.

### 3. Efectividad de los canales de contacto

El canal móvil (`cellular`) presenta una tasa de contratación del 14,74 %, frente al 5,16 % del teléfono fijo (`telephone`).

Esta diferencia sugiere que el canal móvil podría ser especialmente interesante para futuras campañas, aunque sería necesario analizar posibles diferencias entre los clientes contactados por cada canal.

### 4. Frecuencia de contacto

Los clientes contactados una sola vez presentan una tasa de contratación del 13,02 %, mientras que aquellos contactados cuatro o más veces registran una tasa del 7,23 %.

Los resultados muestran una asociación negativa entre el número de contactos durante la campaña y la contratación.

Desde una perspectiva comercial, sería recomendable estudiar la eficiencia de los contactos repetidos y valorar su coste.

### 5. Importancia del historial de campañas

El resultado de campañas anteriores es uno de los factores que presenta mayores diferencias descriptivas.

### 6. Ingresos y variables económicas

Los cuatro grupos de ingresos presentan tasas de contratación similares, situadas entre el 10,84 % y el 11,51 %. Por tanto, los ingresos no parecen ser una variable especialmente diferenciadora cuando se analizan de forma aislada.

Por otro lado, las variables económicas `nr.employed`, `euribor3m` y `emp.var.rate` muestran correlaciones negativas con la contratación, lo que sugiere que el contexto económico podría estar asociado con los resultados de las campañas.
